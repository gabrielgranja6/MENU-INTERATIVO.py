```python
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1P61ZUY2VYOWjpVjV4DAQnOJOYtBmutyWVCSxr2prNBw/export?format=csv&gid=111025361"

df = pd.read_csv(url, sep=";")

df.head()
```

```python
df.columns

```

```python
df["escolaridade"].value_counts(dropna=False)
```

```python
def escolaridade_para_numero(valor):
    """
    Converte o texto da coluna 'escolaridade' em um número padronizado.
    0 = Não informado / desconhecido
    1 = Fundamental incompleto
    2 = Fundamental completo
    3 = Médio incompleto
    4 = Médio completo
    5 = Superior incompleto
    6 = Superior completo
    7 = Pós-graduação / especialização / mestrado / doutorado
    """

    if pd.isna(valor):
        return 0

    texto = str(valor).strip().lower()

    # Fundamental
    if "fundamental" in texto and "incompleto" in texto:
        return 1
    if "fundamental" in texto and "completo" in texto:
        return 2

    # Médio (considerando variação sem acento)
    if ("médio" in texto or "medio" in texto) and "incompleto" in texto:
        return 3
    if ("médio" in texto or "medio" in texto) and "completo" in texto:
        return 4

    # Superior
    if "superior" in texto and "incompleto" in texto:
        return 5
    if "superior" in texto and "completo" in texto:
        return 6

    # Pós, mestrado, doutorado
    if any(p in texto for p in ["pós", "pos", "mestrado", "doutorado", "especialização", "especializacao"]):
        return 7

    return 0
```

# Tratamento da coluna "escolaridade"

Neste notebook, utilizamos um dataset fornecido pelo professor e aplicamos
uma função em Python, via `Series.apply`, para transformar a coluna textual
`escolaridade` em uma coluna numérica `escolaridade_num`.
O objetivo é permitir análises estatísticas a partir de um dado originalmente qualitativo.

```python
df["escolaridade_num"] = df["escolaridade"].apply(escolaridade_para_numero)
df[["escolaridade", "escolaridade_num"]].head(20)



```

```python
## Função de conversão de escolaridade

A função abaixo recebe o texto da escolaridade, padroniza (minúsculas, remoção de espaços)
e, usando `if/elif`, mapeia cada nível para um número entre 0 e 7.
Isso permite comparar os registros numericamente.


# Distribuição da escolaridade numérica
df["escolaridade_num"].value_counts().sort_index()
```

```python
# Média numérica da escolaridade dos respondentes
df["escolaridade_num"].mean()
```

```python
df.to_csv("dados_tratados_escolaridade.csv", index=False)
```

## Conclusão

A transformação da coluna textual `escolaridade` para uma escala numérica
permitiu realizar cálculos e análises que antes não eram possíveis.

A nova coluna `escolaridade_num` mostrou que a maior parte dos respondentes
está entre os níveis 5 e 6 (Superior incompleto / Superior completo).
A média geral foi 5.26, indicando predominância de pessoas com ensino superior.
